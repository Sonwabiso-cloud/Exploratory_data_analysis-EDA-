```python
#import libraries
import pandas as pd
import numpy as np     
import matplotlib.pyplot as plt
import seaborn as sns
# load dataset
df = pd.read_csv('cleaned_stock_prices.csv')
df["date"] = pd.to_datetime(df["date"])
#overview of dataset
df.head()
df.info()
df.describe()

#startistical summary
df[['open', 'close', 'high', 'low']].mean()
df[['open', 'close', 'high', 'low']].median()
df.mode().iloc[0]
df[['open','high', 'low', 'close']].std()
df[['open', 'high', 'low', 'close']].min()  
df[['open', 'high', 'low', 'close']].max()
df[['open', 'high', 'low', 'close']].quantile([0.25, 0.5, 0.75])

#data visualization
# Distribution analysis(histogram)
df[['open', 'high', 'low', 'close']].hist(bins=30, figsize=(10,8))

#volume distribution
plt.figure(figsize=(8,5))
sns.histplot(df['volume'], bins=50)
plt.title('Volume Distribution')    
plt.show()

#box plots
plt.figure(figsize=(10,6))
sns.boxplot(data=df[['open', 'high', 'low', 'close']])
plt.title('Box Plot of Stock Prices')
plt.show()

#scatter plots
plt.figure(figsize=(8,6))
sns.scatterplot(x='open', y='close', data=df)
plt.title('Open vs Close Prices')
plt.show()

#correlation analysis
correlation_matrix = df[['open', 'high', 'low', 'close', 'volume']].corr()
correlation_matrix
#heatmap
plt.figure(figsize=(8,6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.show()

#time series analysis
plt.figure(figsize=(12,6))
plt.plot(df['date'], df['close'], label='Close Price')  
plt.title('Close Price Over Time')
plt.xlabel('Date')      
plt.ylabel('Close Price')  
plt.legend()
plt.show()  

#moving averages
df['MA50'] = df['close'].rolling(window=50).mean()
df['MA200'] = df['close'].rolling(window=200).mean()                
plt.figure(figsize=(12,6))
plt.plot(df['date'], df['close'], label='Close Price')
plt.plot(df['date'], df['MA50'], label='50-Day MA', color='orange')
plt.plot(df['date'], df['MA200'], label='200-Day MA', color='red')
plt.title('Close Price with Moving Averages')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.show()  

# volatility analysis
df['Daily Return'] = df['close'].pct_change()
plt.figure(figsize=(10,5))
sns.histplot(df['Daily Return'].dropna(), bins=50, kde=True)    
plt.title('Daily Return Distribution')
plt.show()

#EDA Validation
df.isna().sum()
df.duplicated().sum()
```
